# Running a Single LCDM Chain

The `sample_joint` function in `cmb_lensing/sample_lcdm.py` draws samples of the LCDM parameters from the joint posterior $P(f, \phi, \theta \mid d)$ using a modified version of the Gibbs sampling algorithm introduced by [Millea et al. 2020](https://arxiv.org/pdf/2002.00965). Here $f$ is the unlensed CMB field, $\phi$ the lensing potential, $\theta$ the LCDM parameters and $d$ the data.

Five parameters can be sampled: $100*\theta_\mathrm{MC}$, $\ln(10^{10} A_s)$, $n_s$, $\Omega_b h^2$ and $\Omega_c h^2$ (`theta_MC_100`, `logA`, `ns`, `ombh2` and `omch2` in the code). The optical depth to reionization `tau` is fixed at 0.05 and the neutrino mass `mnu` at 0.06 eV. The fiducial values of the sampled parameters, their ±5-sigma search ranges and their one-sigma scales are stored as `GROUND_TRUTH`, `PARAM_BOUNDS` and `PARAM_SIGMA` in `cmb_lensing/constants.py`.

## One Gibbs iteration

The Python implementation mirrors CMBLensing.jl's `sample_joint`. Each iteration consists of six steps:

1. **$f$ step** (`gibbs_sample_f`): draw $f$ from its conditional distribution, as a simulated realization plus a Wiener-filter correction.
2. **mix**: change to the mixed parametrization $f^\circ = L(\phi) \cdot D \cdot f$, $\phi^\circ = G \cdot \phi$.
3. **$\phi$ step** (`gibbs_sample_phi`): take one HMC step on the mixed lensing potential $\phi^\circ$, with the mass matrix $G^{-2}(C_\phi^{-1} + N_\phi^{-1})$.
4. **$\theta$ step**: for each sampled parameter, make a Gaussian random-walk Metropolis move of width `proposal_sigmas[name]` on the mixed log-posterior. Proposals outside `param_ranges[name]` or outside the CAMB grid are rejected.
5. **update** $C_\phi$, $C_f$, $D$ and $G$ at the accepted $\theta$.
6. **unmix** back to $(f, \phi)$ with the updated $D$ and $G$.

The $\theta$ step is skipped for the first `num_burn_in_fix_theta` iterations.

## Worked example

The rest of this page samples a single parameter, $\ln(10^{10} A_s)$, from one small temperature map.

```python
from cmb_lensing.sample_lcdm import *
from cmb_lensing.simulate import *
from cmb_lensing.constants import *
from cmb_lensing.camb_spectra import load_sim_cosmology
from scipy.stats import gaussian_kde
```

We begin by simulating a data map at the fiducial LCDM cosmology: a single $64 \times 64$ pixel temperature map with a resolution of $\theta_\mathrm{pix} = 2.5$ arcminutes per pixel and a noise level of $5\,\mu$K-arcmin. The `lmax` argument sets the maximum multipole CAMB is run to. Beyond it, the power spectra are extrapolated in log space.

```python
#generate a data set with the LCDM parameters held at ground truth
ground_truth_params = load_sim_cosmology(GROUND_TRUTH)
noise_level = 5
data_set = load_sim(nside = 64, theta_pix = 2.5, pol = "I", 
                    master_seed = 42, uk_arcmin_t = noise_level,
                    **ground_truth_params, r = 0, 
                    nt = 0, lmax = 4000)
```

Next we run `sample_joint`, which takes the data set as input and returns samples of the cosmological parameters. Here we sample only $\ln(10^{10} A_s)$ and hold the other four parameters at their fiducial values, but `should_sample` can switch on any subset of the five to sample them jointly. In general, the more parameters are sampled jointly, the stronger the degeneracies between them become, and larger maps with finer resolution are needed to obtain meaningful constraints. The chain starts at the lower edge of the search range, far from the ground truth, so that the burn-in is easy to see.

The first time you run the following cell, the ~7.9 GB `camb_grid_spline.npz` file is downloaded automatically from Hugging Face into the `camb_splines/` directory (see [Installation](../getting-started/installation/#downloading-the-camb-grid)), so expect a wait. The two progress plots are written to `cmb_lensing/sample_lcdm_output/`, which must exist before the cell is run.

```python
#Initialize logA to the total other side of the search range (1-sigma below fiducial)
param_init = dict(GROUND_TRUTH)
param_init["logA"] = PARAM_BOUNDS["logA"][0]

#The 1-D Metropolis Hastings proposal width for each parameter you want to sample
proposal_sigmas = {}
proposal_sigmas["logA"] = 1e-2

#Whether or not to sample each of the LCDM parameters
should_sample = {}
should_sample["logA"] = True
should_sample["ns"] = False
should_sample["ombh2"] = False
should_sample["omch2"] = False
should_sample["theta_MC_100"] = False

#Whether or not to log certain diagnostic statistics
advanced_logging = {}
advanced_logging["phi_acceptance"] = False
advanced_logging["lcdm_acceptance"] = False
advanced_logging["plot_log_pdf"] = True
advanced_logging["plot_lcdm_sigmas"] = True

#Run the sampling algorithm
param_vals, _, _ = sample_joint(data_set, param_init, proposal_sigmas, PARAM_BOUNDS, 
                                should_sample, noise_level, advanced_logging, 
                                fixed_fields = False, phi_init = "ZEROES",
                                iters_per_chain = 300, num_burn_in_fix_theta = 0, 
                                seed = 67)
```

## Outputs

`sample_joint` returns `(param_vals, phi, sub_key)`: a dictionary holding one list of samples per parameter, the final lensing potential and the final random key. When `hpc_path` is set, every iteration also appends the latest value of each sampled parameter to `{hpc_path}{theta}_map_{map_idx}_chain_{sub_chain_idx}_history.txt`. These history files are what `cmb_lensing/chain_analysis.py` reads (see [Analyzing Chain Data](../chain_analysis_tutorial/)).

## Analyzing the samples

With the samples in hand, the next step is to analyze them. The first plot below is the progress plot that `sample_joint` writes: the trace of the $\ln(10^{10} A_s)$ samples minus the ground truth value, in units of the parameter's `PARAM_SIGMA`. There is an initial burn-in because the chain starts far away from the typical set. From around iteration 200 the chain samples from the posterior of $\ln(10^{10} A_s)$, and the normalized samples scatter within about one sigma of the ground truth.

The second plot is the log-probability of the distribution being sampled. It plateaus even sooner than the parameter trace does, which indicates that the samples of the unlensed field and the lensing potential have a much larger influence on the log-probability than the value of $\ln(10^{10} A_s)$ does.

```python
#plot the logpdf history and the logA traces
from IPython.display import Image, display
HERE = os.getcwd()
ONE_UP = os.path.dirname(HERE)
IMAGE_FOLDER = os.path.join(ONE_UP, "cmb_lensing", "sample_lcdm_output")
display(Image(filename = os.path.join(IMAGE_FOLDER, "lcdm_progress.png"), width = 800))
display(Image(filename = os.path.join(IMAGE_FOLDER, "logpdf_progress.png"), width = 800))
```

We can also plot the un-normalized trace of the $\ln(10^{10} A_s)$ samples. The story is essentially the same: the chain starts far from the ground truth, burns in, and then samples from the posterior of $\ln(10^{10} A_s)$. Notice that the samples do not settle on the ground truth value but scatter around it. This is expected: the goal is to learn the distribution of $\ln(10^{10} A_s)$, not a single value. The scatter of the samples tells us how well the data constrain the parameter: a large scatter means a weak constraint, and a small scatter a tight one.

```python
# --------- Trace Plot -----------------------------------------------------------------------------------------------------------------------
%matplotlib inline
plt.figure(figsize = (8, 5))
plt.plot(np.arange(len(param_vals["logA"])), param_vals["logA"], marker = "o", color = "steelblue", label = "LogA Samples")
plt.axhline(GROUND_TRUTH["logA"], color = "black", label = "Ground Truth LogA")
plt.xlabel("Sample Number")
plt.ylabel("LogA Value")
plt.title("Sample Trace of LogA")
plt.legend()
plt.show()
```

We can also turn the samples into a probability density. Before anything else we discard the first 175 samples, which are not in the typical set. The cell below then uses SciPy's `gaussian_kde`, a quick way to look at a single chain. The package's own analysis (`cmb_lensing/chain_analysis.py`) does not use a kernel density estimate: it assumes that each map's posterior is a normal distribution with the mean and standard deviation of that map's samples.

In practice the samples should also be thinned by their integrated autocorrelation time first, but this run is too short for that to leave enough samples to form a distribution. In general, thinning widens the learned distribution, so the estimate shown here is overconfident relative to what it should be.

```python
#-------- Gaussian KDE ------------------------------------------------------------------------------------------------------------------------
burn_in = 175
burnt_samples = param_vals["logA"][burn_in:]
grid = np.linspace(PARAM_BOUNDS["logA"][0], PARAM_BOUNDS["logA"][-1], 1000)
dx = np.diff(grid)
log_a_pdf = gaussian_kde(burnt_samples)(grid)

log_a_mode = grid[np.argmax(log_a_pdf)]
log_a_mean = np.dot(log_a_pdf, grid * dx[0])
log_a_std = np.sqrt(np.sum(dx[0] * log_a_pdf * (grid - log_a_mean)**2))

plt.figure(figsize = (8, 5))
plt.plot(grid, log_a_pdf, label = "LogA KDE", color = "steelblue")
plt.axvline(GROUND_TRUTH["logA"], color = "black", label = f"Ground Truth = {round(GROUND_TRUTH["logA"], 4)}")
plt.axvline(log_a_mode, color = "indianred", label = f"Mode = {round(log_a_mode, 4)}")
plt.axvline(log_a_mean, color = "green", label = f"Mean = {round(log_a_mean, 4)}")
plt.axvline(log_a_mean - log_a_std, color = "grey", label = f"Std = {round(log_a_std, 4)}")
plt.axvline(log_a_mean + log_a_std, color = "grey")
plt.plot([], label = f"Z-Score = {(log_a_mean - GROUND_TRUTH["logA"])/log_a_std}", color = "purple")
plt.xlabel("LogA Value")
plt.ylabel("Probability Density")
plt.title("Learned Distribution for LogA (UN-thinned Samples)")
plt.legend()
plt.show()
```

Below we plot the autocorrelation function against lag, which tells us how correlated two samples are when they are a given number of iterations apart. Summing the autocorrelation function gives a single number, the integrated autocorrelation time (IAT): roughly how many iterations separate two independent samples. Here it turns out to be very small, only about 4. The two functions defined in the cell are also available as `autocorrelation` and `integrated_autocorrelation_time` in `cmb_lensing/mcmc_diagnostics.py`.

```python
#-------- Autocorrelation Function ------------------------------------------------------------------------------------------------------------
def autocorrelation(chain, max_lag = None):
    """Normalized autocorrelation of a 1D chain of samples."""
    n = len(chain)
    if max_lag is None:
        max_lag = n // 2
    x = chain - np.mean(chain)
    variance = np.var(chain)
    acf = np.correlate(x, x, mode = "full")[n - 1:]  #positive lags only
    acf = acf[:max_lag] / (variance * n)
    return acf

#-------- Integrated Autocorrelation Time -----------------------------------------------------------------------------------------------------
def integrated_autocorrelation_time(acf, c = 5.0):
    """Sokal's automatic windowing — truncate at lag M where M < c * tau(M)."""
    tau = 1.0
    for m in range(1, len(acf)):
        tau += 2.0 * acf[m]
        if m >= c * tau:
            return tau
    return tau

acf = autocorrelation(burnt_samples)
iat = np.ceil(integrated_autocorrelation_time(acf))
plt.figure(figsize = (8, 5))
plt.plot(acf, color = "steelblue", label = f"Autocorrelation (IAT = {iat})")
plt.axhline(0, color = "black", label = "Zero Line")
plt.legend()
plt.ylabel("LogA Autocorrelation")
plt.xlabel("Lag")
plt.title("ACF v.s. Lag")
plt.show()
```

Here we have drawn samples of a single parameter from a single chain on a single data map. One chain on one map is not a statistically powerful statement, because of MCMC noise and cosmic variance (especially on such a small map). The standard workflow is to run the sampler on many data map realizations at the ground truth cosmology (of order 10 to 100) and, for each map, several sub-chains (typically 5 to 10) that each start from random values of the LCDM parameters.

The analysis then thins every sub-chain by its integrated autocorrelation time, pools the samples of each map, and summarizes each map by a normal distribution with the mean and standard deviation of its samples. The product of these normal distributions over the maps is a single distribution: the best estimate of the parameter given all of the data. The machinery for runs like this is in `get_lcdm_samples/` (see [LCDM Sampling on the HPC](../pipelines/lcdm-chains/)), and [Analyzing Chain Data](../chain_analysis_tutorial/) walks through the analysis.

## Tuning

Before a large experiment, run a pilot chain on a small map (`python -m cmb_lensing.sample_lcdm` runs one) and tune

- `proposal_sigmas` toward an acceptance rate of about 44% per parameter,
- `hmc_step_size` and `hmc_num_steps` toward an acceptance rate of about 65% for $\phi$.

`advanced_logging = {"phi_acceptance": True, "lcdm_acceptance": True, ...}` prints the running acceptance rates, and `plot_log_pdf` / `plot_lcdm_sigmas` write the progress plots shown above.
